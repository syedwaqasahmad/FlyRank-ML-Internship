## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Decision this improves: which pages are at highest risk of "churning" (losing search visibility) so a reviewer can intervene before it happens, not after.

Who acts on it: a content/SEO reviewer with limited time, deciding which pages to check first.

Cost of a wrong call: a false positive wastes review time on a page that was never at risk; a false negative means a page silently churns out of visibility and the reviewer finds out too late to act.

```python
# Confirming the churn signal we're using is real and present in the data
print("Churn signal used:", "trend_direction == 'down'")
print("Values in trend_direction:", df["trend_direction"].unique())
```

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

Using the real FlyRank starter data, I looked at how many pages have actually "churned" (declined in search visibility), and how many show the early warning signs of churn risk (stale + still getting traffic) before checking whether those warning signs actually line up with real declines.

```python
churned = df["trend_direction"].str.lower().eq("down")
at_risk = (df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 100)

print(f"Pages that have 'churned' (declining): {churned.sum()} of {len(df)} ({churned.mean()*100:.1f}%)")
print(f"Pages stale + still getting traffic (churn-risk candidates): {at_risk.sum()} of {len(df)} ({at_risk.mean()*100:.1f}%)")
print(f"Of the churn-risk candidates, how many actually churned: {(at_risk & churned).sum()} of {at_risk.sum()}")
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

What I can claim: this is an observed, directional signal — pages that are stale and still getting search visibility "churn" (decline) at a noticeably higher rate (74%) than the overall base rate (54.2%). That's evidence worth building on, not proof of a mechanism.

What I can't claim: this doesn't prove causation — I can't say staleness causes decline, only that they're correlated in this data. I'm also not predicting Google's algorithm; this supports a human reviewer's prioritization decision, it doesn't replace their judgment.
